3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

1.

```python
result_matrix = np.dot(B, A)
result_matrix
```

2. Результатом множення матриці розміром (2, **2**) на (**2**, 1) буде структура (матриця/вектор) розміром (2, 1)

```python
result_matrix_mult_v = np.dot(result_matrix, C)
result_matrix_mult_v, result_matrix_mult_v.shape
```

3. Результатом множення матриці розміром (2, **2**) на (**2**, 2) буде структура (матриця) розміром (2, 2)

```python
np.dot(result_matrix, D)
```

Для вектора vector операція .T не змінює його структуру, оскільки він має розмірність (10,) тобто є одновимірним
